```python
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()

model =  ChatGroq(
        model="openai/gpt-oss-20b",
        temperature=0.2
    )
```

```python
class BlogState(TypedDict):

    title: str
    outline: str
    content: str
    score: str
```

```python
def create_outline(state: BlogState) -> BlogState:

    # fetch title
    title = state['title']

    # call llm gen outline
    prompt = f'Generate a detailed outline for a blog on the topic - {title}'
    outline = model.invoke(prompt).content

    # update state
    state['outline'] = outline

    return state
```

```python
def create_blog(state: BlogState) -> BlogState:

    title = state['title']
    outline = state['outline']

    prompt = f'Write a detailed blog on the title - {title} using the follwing outline \n {outline}'

    content = model.invoke(prompt).content

    state['content'] = content

    return state
```

```python
def evaluate(state: BlogState) -> BlogState:

    title = state['title']
    outline = state['outline']
    content = state['content']

    prompt = f'Evaluate the following blog content - {content} based on the follwing outline \n {outline} and provide a final score out of 10'

    score = model.invoke(prompt).content

    state['score'] = score

    return state
```

```python
graph = StateGraph(BlogState)

# nodes
graph.add_node('create_outline', create_outline)
graph.add_node('create_blog', create_blog)
graph.add_node('evaluate', evaluate)

# edges
graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline', 'create_blog')
graph.add_edge('create_blog', 'evaluate')
graph.add_edge('evaluate', END)

workflow = graph.compile()
```

```python
intial_state = {'title': 'Rise of short form cricket in india'}

final_state = workflow.invoke(intial_state)

print(final_state)
```
